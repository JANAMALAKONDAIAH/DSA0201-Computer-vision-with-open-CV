```python
import cv2

image_path = '/content/MSD07.jpg'
image = cv2.imread(image_path)

if image is not None:
    height, width, _ = image.shape
    print(f"Image dimensions using OpenCV: {width}x{height} pixels")
else:
    print(f"Error: Could not load image from {image_path}")
```

```python
if image is not None:
    _, _, channels = image.shape
    print(f"Number of channels: {channels}")
else:
    print("Image not loaded. Please ensure the image path is correct and the image is loaded.")
```

```python
import matplotlib.pyplot as plt

if image is not None:
    # Convert the image to grayscale
    gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

    # Display the grayscale image
    plt.imshow(gray_image, cmap='gray')
    plt.title('Grayscale Image')
    plt.axis('off') # Hide axes ticks
    plt.show()
else:
    print("Image not loaded. Cannot convert to grayscale.")
```

```python
if gray_image is not None:
    # Resize the grayscale image to 100x100
    resized_image = cv2.resize(gray_image, (100, 100))

    # Display the resized image
    plt.imshow(resized_image, cmap='gray')
    plt.title('Resized Grayscale Image (100x100)')
    plt.axis('off')
    plt.show()
else:
    print("Grayscale image not loaded. Cannot resize.")
```

```python
if gray_image is not None:
    # Resize the grayscale image to 200x200
    resized_image_200x200 = cv2.resize(gray_image, (200, 200))

    # Display the resized image
    plt.imshow(resized_image_200x200, cmap='gray')
    plt.title('Resized Grayscale Image (200x200)')
    plt.axis('off')
    plt.show()
else:
    print("Grayscale image not loaded. Cannot resize.")
```

```python
if gray_image is not None:
    # Convert grayscale image to binary using a threshold
    # The threshold value (127) can be adjusted. Pixels above 127 become 255 (white), below become 0 (black).
    # THRESH_BINARY_INV would invert this.
    _, binary_image = cv2.threshold(gray_image, 127, 255, cv2.THRESH_BINARY)

    # Display the binary image
    plt.imshow(binary_image, cmap='gray')
    plt.title('Binary Image')
    plt.axis('off')
    plt.show()
else:
    print("Grayscale image not loaded. Cannot convert to binary.")
```

```python
if gray_image is not None:
    # Apply a threshold of 180 to convert grayscale to binary
    # Pixels with intensity > 180 become 255 (white), otherwise 0 (black)
    _, binary_image_thresh180 = cv2.threshold(gray_image, 180, 255, cv2.THRESH_BINARY)

    # Display the binary image
    plt.imshow(binary_image_thresh180, cmap='gray')
    plt.title('Binary Image (Threshold = 180)')
    plt.axis('off')
    plt.show()
else:
    print("Grayscale image not loaded. Cannot apply threshold.")
```

```python
if gray_image is not None:
    # Apply Canny edge detection
    # The threshold1 and threshold2 values (e.g., 100, 200) can be adjusted
    edges = cv2.Canny(gray_image, 100, 200)

    # Display the edges
    plt.imshow(edges, cmap='gray')
    plt.title('Edge Detection (Canny)')
    plt.axis('off')
    plt.show()
else:
    print("Grayscale image not loaded. Cannot detect edges.")
```
